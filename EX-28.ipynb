```python
import re

passwords = [
    "123456",
    "admin",
    "password",
    "Welcome123",
    "Strong@123"
]

print("Weak Password Detection\n")

for pwd in passwords:
    if len(pwd) < 8 or pwd.isalpha() or pwd.isdigit():
        print(pwd, "-> Weak")
    else:
        print(pwd, "-> Strong")

print("\nSQL Injection Detection")

payloads = [
    "' OR 1=1 --",
    "admin",
    "UNION SELECT *",
    "../etc/passwd",
    "normal search"
]

for p in payloads:

    if re.search(r"or\s+1=1|union|--", p, re.I):
        print(p, "-> SQL Injection")

    elif "../" in p:
        print(p, "-> Directory Traversal")

    else:
        print(p, "-> Normal Request")
```
